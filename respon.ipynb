6. Lakukan rotasi searah dan berlawanan jarum jam terhadap citra grayscale Gambar B 
dengan derajat rotasi sesuai dengan 3 digit terakhir NIM. Tampilkan 2 citra hasil rotasi 
tersebut dalam subplot 1x2. 

```python
def rotasi_bebas_crop(image, derajat):
    h, w = image.shape
    sudut = np.radians(derajat)

    # Titik pusat gambar
    cx, cy = w / 2, h / 2

    # Canvas output ukuran SAMA dengan input
    hasil = np.zeros((h, w), dtype=image.dtype)

    for i in range(h):
        for j in range(w):
            # Inverse mapping
            x = (j - cx) * np.cos(-sudut) - (i - cy) * np.sin(-sudut) + cx
            y = (j - cx) * np.sin(-sudut) + (i - cy) * np.cos(-sudut) + cy

            x, y = int(round(x)), int(round(y))

            # Kalau masih dalam batas → salin piksel
            # Kalau di luar batas → biarkan 0 (hitam/terpotong)
            if 0 <= x < w and 0 <= y < h:
                hasil[i, j] = image[y, x]

    return hasil
```

Fungsi ini mengimplementasikan teknik rotasi citra manual menggunakan metode inverse mapping tanpa mengubah ukuran canvas asli. Dengan menghitung posisi piksel sumber berdasarkan koordinat pusat gambar dan sudut rotasi yang diberikan, fungsi ini memindahkan nilai piksel ke posisi baru; area yang berada di luar jangkauan dimensi gambar asli setelah diputar akan secara otomatis terpotong (cropped) dan diisi dengan warna hitam (nilai 0).

```python
searahJarumJam = rotasi_bebas_crop(gambarBGray, 70)
berlawananArahJarumJam = rotasi_bebas_crop(gambarBGray, -70)


plt.subplot(1, 2, 1)
plt.imshow(searahJarumJam, cmap='gray')
plt.title('searahJarumJam')
plt.subplot(1, 2, 2)
plt.imshow(berlawananArahJarumJam, cmap='gray')
plt.title('berlawananArahJarumJam')
```

Kode ini menerapkan fungsi rotasi pada gambarBGray dengan sudut sebesar $70^\circ$ (NIM SAYA)untuk menghasilkan efek putaran searah dan berlawanan arah jarum jam. Menggunakan fungsi rotasi_bebas_crop, citra diputar di sekitar titik pusatnya, lalu hasilnya ditampilkan berdampingan dalam satu baris untuk membandingkan orientasi visual yang dihasilkan. Karena ukuran canvas tetap dipertahankan, bagian sudut citra yang keluar dari batas dimensi asli setelah rotasi akan terlihat terpotong.

7. Lakukan thresholding terhadap citra grayscale Gambar B dengan nilai threshold sesuai 
dengan 3 digit terakhir NIM. Tampilkan citra grayscale dan citra thresholding dalam 
subplot 1x2. 

```python
def tresholding(img, treshold):
    h, w = img.shape 
    rot = np.zeros((h, w), dtype=img.dtype)
    for i in range(h): 
        for j in range(w): 
            if( img[i, j] > treshold):
                rot[i, j] = 255
            else :
                rot[i, j] = 0
    return rot
```

Fungsi ini mengimplementasikan teknik thresholding manual (pengambangan) untuk mengubah citra grayscale menjadi citra biner (hitam-putih murni).Secara teknis, kode ini bekerja dengan cara:Iterasi Piksel: Memeriksa setiap piksel pada koordinat $(i, j)$ di seluruh dimensi gambar.Logika Pemutusan: Membandingkan nilai intensitas piksel dengan nilai treshold yang ditentukan.Segmentasi: Jika nilai piksel lebih besar dari ambang batas, piksel diubah menjadi putih (255); jika lebih kecil atau sama, diubah menjadi hitam (0).Hasil akhir dari fungsi ini adalah gambar dengan kontras tinggi yang biasanya digunakan untuk memisahkan objek utama dari latar belakangnya.

```python
gambar = tresholding(gambarBGray, 70)

plt.figure(figsize=(10,7))

plt.subplot(1,2,1)
plt.imshow(gambarBGray, cmap='gray')
plt.title('Asli')
plt.subplot(1,2,2)
plt.imshow(gambar, cmap='gray')
plt.title('Thresholding')
```

Kode ini menerapkan fungsi thresholding pada gambarBGray dengan nilai ambang batas (threshold) sebesar 70. Hasilnya ditampilkan berdampingan dengan gambar asli untuk membandingkan perbedaan visualnya.

Melalui proses ini, setiap bagian gambar yang memiliki intensitas cahaya di atas 70 akan dikonversi menjadi putih murni, sedangkan sisanya menjadi hitam murni. Visualisasi ini sangat berguna untuk melihat bagaimana nilai ambang batas tersebut berhasil memisahkan objek dari latar belakangnya secara biner.

8. Dengan menggunakan citra grayscale pada gambarB, lakukan: 

a. 
Buat versi citra grayscale yang sudah ditambahkan zero padding sebesar 150 piksel di 
semua arah (kanan, kiri, atas, bawah). 

```python

def zeroPadding(img, padding):
    h, w = img.shape[:]

    hasil = np.zeros((h + padding, w+ padding), dtype=img.dtype)
    gambar = img + hasil
    return gambar
```

```python
def zero_padding(citra, pad):
    tinggi, lebar = citra.shape
    
    hasil = np.zeros((tinggi + 2*pad, lebar + 2*pad), dtype=citra.dtype)
    
    hasil[pad:pad+tinggi, pad:pad+lebar] = citra
    
    return hasil
```

Fungsi ini mengimplementasikan teknik Zero Padding, yaitu menambahkan bingkai berisi nilai nol (hitam) di sekeliling citra asli. Secara teknis, fungsi ini bekerja dengan cara:

Ekspansi Dimensi: Membuat matriks baru menggunakan np.zeros dengan ukuran yang diperluas sebesar 2 * pad pada sisi tinggi dan lebar.

Penempatan Citra: Menyalin citra asli ke bagian tengah matriks baru tersebut, tepatnya mulai dari indeks koordinat pad hingga pad + dimensi_asli.

```python
def timpa (image1, image2):
    h, w = image1.shape
    hasil = image1.copy()
    for i in range(h):
        for j in range(w):
            if image2[i, j] > 0:
                hasil[i, j] = image2[i, j]
    return hasil
```

Cara kerjanya adalah dengan menyalin image1 sebagai latar belakang, lalu menimpanya dengan piksel dari image2 hanya jika piksel tersebut memiliki nilai di atas 0. Dengan kata lain, warna hitam pada gambar kedua dianggap "transparan", sehingga objek non-hitam akan muncul menempel di atas gambar pertama tanpa menutupi seluruh permukaannya.

```python
bPadding = zero_padding(gambarBGray, 150)

gambarRot = rotasi_bebas_crop(bPadding, 30)
gambarTimpa1 = timpa(bPadding, gambarRot)


gambarRot2 = rotasi_bebas_crop(bPadding, -30)
gambarTimpa2 = timpa(gambarTimpa1, gambarRot2)

plt.figure(figsize=(12, 12))
plt.subplot(1, 3, 1)
plt.imshow(bPadding, cmap='gray')

plt.subplot(1, 3, 2)
plt.imshow(gambarTimpa1, cmap='gray')
plt.subplot(1, 3, 3)
plt.imshow(gambarTimpa2, cmap='gray')
```

Proses dimulai dengan Zero Padding 150 piksel supaya pas gambar diputar, nggak ada bagian yang kepotong sama sekali. Setelah itu, gambar diputar dua kali—sekali ke kanan ($30^\circ$) dan sekali ke kiri ($-30^\circ$). Dengan fungsi timpa, kedua hasil rotasi itu ditumpuk bertahap ke atas gambar asli. Hasil akhirnya, kamu dapet satu citra yang punya efek "bayangan" atau objek yang mengganda secara simetris, tapi tetep rapi karena ruang padding yang luas tadi. Efeknya dapet banget tanpa harus pakai teknik yang ribet.
